API utilizada: **JSONPlaceholder**  
Documentação: https://jsonplaceholder.typicode.com/  
Guia de uso: https://jsonplaceholder.typicode.com/guide/

A JSONPlaceholder é uma API pública de testes. Ela permite simular requisições `GET`, `POST`, `PUT`, `PATCH` e `DELETE` sem precisar de login ou cadastro.

> Observação importante: os dados criados, atualizados ou excluídos são apenas simulados. A API responde como se a operação tivesse acontecido, mas não altera a base real.
